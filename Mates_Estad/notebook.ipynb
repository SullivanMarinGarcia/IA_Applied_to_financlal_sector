```python
import numpy as np

A = np.array([
    [7, 2, -1],
    [0, 3, 5],
    [4, 2, 2],
    [1, 1, -1]
])

B = np.array([
    [1, 1, 1],
    [2, 2, 2],
    [3, 3, 3],
    [4, 4, 4]
])
```

```python
suma = A + B
resta = A - B

print(suma)
print(resta)
```

```python
producto = A * B
print(producto)
```

```python
C = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])  # Dimensión (3, 2)

producto_matricial = A @ C  # Resultado: matriz de dimensión (4, 2)

print(producto_matricial)
```

Ejercicio de prueba, todos son buenos, al cabo de un año como queda la cosa?

```python
matriz_transicion = np.array([[0.9, 0.4], [0.1, 0.6]])
matriz_estados = np.array([1, 0])

primer_mes = matriz_transicion @ matriz_estados
print(primer_mes)

primer_year = np.linalg.matrix_power(matriz_transicion, 12) @ matriz_estados
print(primer_year)
```

```python
P = np.array([[0.9, 0.15, 0.25], [0.075, 0.8, 0.25], [0.025, 0.05, 0.5]])
w = np.array([0, 0, 1])
w_prima_1 = np.array([1, 0, 0])
w_prima_2 = np.array([0, 1, 0])

fourth_week = np.linalg.matrix_power(P, 4) @ w
print(fourth_week)

test1 = np.linalg.matrix_power(P, 100) @ w
test2 = np.linalg.matrix_power(P, 100) @ w_prima_1
test3 = np.linalg.matrix_power(P, 100) @ w_prima_2

print(test1)
print(test2)
print(test3)
```
