### Variables *discretas*

#### Ejemplo 1: Variable que toma valores entre 0 y 7 con igual probabilidad

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simulamos valores
resultado_dado = np.random.randint(1,7,10000)
```

```python
# Histograma de valores
plt.hist(resultado_dado, bins = np.arange(9)-.5, density = True)
plt.show()
```


**Ejercicio práctico:**  
Una empresa lanza un producto con éxito en el 60% de los casos.

1. ¿Cuál es la probabilidad de obtener 8 éxitos en 12 lanzamientos?
2. ¿Cuál es la probabilidad de obtener 7 o más éxitos en 12 lanzamientos?


```python
from scipy.stats import binom

# 1)
prob = binom.pmf(k = 8, n = 12, p = 0.6)
print(f"Probabilidad de 8 éxitos: {prob:.3f}")
```

```python
# 2)
prob = 1 - binom.cdf(k = 6, n = 12, p = 0.6)
print(f"Probabilidad de 7 o más éxitos: {prob:.3f}")
```
