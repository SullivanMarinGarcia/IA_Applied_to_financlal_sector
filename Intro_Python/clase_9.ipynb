### Ejemplo de uso del `chunker` con Pandas para procesar un archivo grande

Cuando trabajamos con archivos CSV muy grandes, intentar cargarlos completamente en memoria puede causar problemas de desbordamiento de memoria (`MemoryError`). El parámetro `chunksize` en `pd.read_csv` nos permite leer el archivo en fragmentos (chunks), procesándolos uno por uno.

```python
# Creamos un 'chunker' para leer el archivo 'big_file.csv' en fragmentos de 1000 filas.
# El 'chunker' es un iterador, lo que significa que no carga todo el archivo a la vez.
chunker = pd.read_csv('./big_file.csv', chunksize=1000)

print("Iterando a través de los chunks y mostrando la forma de cada uno:")
for i, piece in enumerate(chunker):
    print(f"Chunk {i+1}: {piece.shape} (filas, columnas)")

# Reiniciamos el chunker si queremos iterar de nuevo desde el principio
# chunker = pd.read_csv('./big_file.csv', chunksize=1000)
```

### Problemática: Concatenar todos los chunks en memoria

Una tentación común es intentar concatenar todos los chunks para formar un único DataFrame grande. Si el archivo es **demasiado grande**, esto anulará el propósito de `chunksize` y podría seguir causando `MemoryError`.

Aunque para `big_file.csv` (10,000 filas) esto no es un problema, imagina un archivo con billones de filas.

```python
# NO RECOMENDADO para archivos EXTREMADAMENTE GRANDES debido al consumo de memoria.
# Sin embargo, lo mostramos como ejemplo de un patrón que puede ser problemático.

# Reiniciamos el chunker
chunker_for_concat = pd.read_csv('./big_file.csv', chunksize=1000)

all_chunks = []
for piece in chunker_for_concat:
    all_chunks.append(piece)

# Esto podría fallar para archivos gigantes
df_full = pd.concat(all_chunks, ignore_index=True)

print("Forma del DataFrame resultante después de concatenar todos los chunks:", df_full.shape)
df_full.head()
```

### Solución eficiente: Procesar y agregar chunks sin cargar todo en memoria

La forma correcta de trabajar con archivos muy grandes es procesar cada chunk individualmente y acumular los resultados de forma agregada, en lugar de intentar construir un único DataFrame gigante en memoria.

Por ejemplo, si queremos calcular la suma de una columna numérica, podemos sumar esa columna en cada chunk y luego sumar todas las sumas parciales.

```python
# Reiniciamos el chunker
chunker_for_agg = pd.read_csv('./big_file.csv', chunksize=1000)

# Lista para almacenar los resultados agregados de cada chunk
sum_of_one_column = []

print("Calculando la suma de la columna 'one' para cada chunk...")
for i, piece in enumerate(chunker_for_agg):
    chunk_sum = piece['one'].sum()
    sum_of_one_column.append(chunk_sum)
    print(f"Suma de 'one' en Chunk {i+1}: {chunk_sum:.2f}")

# Sumamos todos los resultados parciales para obtener la suma total
total_sum = sum(sum_of_one_column)

print(f"\nSuma total de la columna 'one' en todo el archivo: {total_sum:.2f}")
```

Este enfoque evita cargar el archivo completo en memoria, haciendo posible procesar conjuntos de datos que de otro modo serían inmanejables.

## Conversión entre datimes y strings

El módulo datetime tiene dos métodos para pasar strings a fechas y viceversa:

* __strptime:__ de string a DateTime object

* __strftime:__ de DateTime object a string

__String a DateTime__

```python
# La parte "%d/%m/%Y" explica el formato de la fecha "08/04/2021"
fecha = datetime.strptime('08-04-2021', '%d-%m-%Y')
```

```python
fecha
```

__DateTime a String__

```python
# El string "%d/%m/%Y" especifica el formato deseado
datetime.strftime(fecha, '%d/%m/%Y')
```

__String to String__ en formato diferente

```python
# Parte 1: Crea objeto DateTime desde String   |   Parte 2: Convierte DateTime a String
datetime.strptime('08-04-2021', '%d-%m-%Y').strftime('%d/%m/%Y')
```

# Time series en Pandas

- Pandas tiene los objetos ``Timestamp`` que combinan la facilidad de  ``datetime`` y ``dateutil`` con una implementación vectorizada eficiente.

- Usando los objetos ``Timestamp``, pandas construye un ``DatetimeIndex``  para ser usado como índice en ``Series`` o ``DataFrame``.

- Las clases de Pandas para series temporales son:

    * ``Timestamp``: reemplazo de ``datetime``, basado en el sistema más eficiente de ``numpy.datetime64``. La estructura del índice asociada será ``DatetimeIndex``.
    * ``Period``: para periodos de tiempo. La estructura del índice asociado es ``PeriodIndex``.
    * ``Timedelta``: para variación de tiempo o duración (más eficiente que ``datetime.timedelta``). La estructura del índice asociado es ``TimedeltaIndex``.

- Los tipos más básicos son ``Timestamp`` y ``DatetimeIndex``.

- La forma más común de crear objetos de este tipo es usando la función ``pd.to_datetime()`` .

- Puede parsear una gran variedad de formatos.

- Pasar una fecha a esta función produce un ``Timestamp``; pasar una serie de fechas por defecto devuelve un ``DatetimeIndex``:

- Las frecuencias mensual, trimestral y anual se proporcionan como fecha al final del periodo especificado.

- Además se puede cambiar el mes usado en frecuencias trimestrales o anuales añadiendo un código de tres letras al final:
 * ``Q-JAN``, ``BQ-FEB``, ``QS-MAR``, ``BQS-APR``, etc.

### Facilidades de fechas en pandas:

Leemos el fichero con datos de niveles de NO $_{2}$

```python
air_quality = pd.read_csv("./air_quality_no2_long.csv", parse_dates=["datetime"])
air_quality.head()
```

```python
air_quality.city.unique()
```

Podemos asignar cada fila a un mes o a la semana etc

```python
air_quality["month"] = air_quality["datetime"].dt.month
air_quality["week"] = air_quality["datetime"].dt.isocalendar().week
air_quality.head()
```

Podemos calcular estadísticas basadas en la fecha, por ejemplo calcular datos por día de la semana

```python
air_quality.groupby(
    [air_quality["datetime"].dt.weekday, "location"])["value"].mean()
```

Podemos dibujar la concentración de NO $_{2}$ para cada hora

```python
fig, axs = plt.subplots(figsize=(12, 4))

air_quality.groupby(air_quality["datetime"].dt.hour)["value"].mean().plot(
    kind='bar', rot=0, ax=axs)

plt.xlabel("Hour of the day");  # custom x label using matplotlib

plt.ylabel("$NO_2 (µg/m^3)$");

plt.title('Concentración NO2')
```

### Time Series con índices duplicados

```python
dates = pd.DatetimeIndex(['1/1/2000', '1/2/2000', '1/2/2000',
                          '1/2/2000', '1/3/2000'])
dup_ts = pd.Series(np.arange(5), index=dates)
dup_ts
```

```python
dup_ts.index.is_unique
```

```python
dup_ts['1/2/2000']  # duplicated
```

- Para eliminarlos podemos agrupar por nivel 0 y realizar una operación que los unifique.

```python
grouped = dup_ts.groupby(level=0)
grouped.mean()
```

```python
grouped.first()
```

```python
grouped.count()
```
